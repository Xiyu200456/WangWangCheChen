**Dataset 1: Ember China Solar PV Export**

```python
import pandas as pd
from pathlib import Path
```

```python
# Import Ember Solar PV Export Dataset
data_path = Path("../data/mart_solar_exports_full_release_monthly.csv")

solar_exports = pd.read_csv(data_path)

# Report dataset dimensions
print("Number of rows:", solar_exports.shape[0])
print("Number of columns:", solar_exports.shape[1])
```

```python
# Check temporal coverage
print("Earliest date:", solar_exports["Date"].min())
print("Latest date:", solar_exports["Date"].max())
```

```python
# Check geographic coverage
print("Number of unique areas:", solar_exports["Area"].nunique())

print("\nRegions:")
print(solar_exports["Region"].value_counts(dropna=False))
```

```python
# Check product categories
print(solar_exports["Commodity type"].value_counts())

print("\nArea types:")
print(solar_exports["Area type"].value_counts())
```

```python
# Import IRENA Solar PV Capacity Dataset
irena_path = Path("../data/IRENA.xlsx")

irena_raw = pd.read_excel(
    irena_path,
    sheet_name="ELEC-C",
    header=None,
    skiprows=2,
    usecols="A:F",
    names=[
        "Country/area",
        "Technology",
        "Data Type",
        "Grid connection",
        "Year",
        "Value"
    ],
    na_values=[".."]
)

irena = irena_raw.loc[irena_raw["Year"].notna()].copy()
```

```python
## Print Result
print("Number of rows:", irena.shape[0])
print("Number of columns:", irena.shape[1])
irena.head()
```

```python
# Display dataset information
irena.info()
```

```python
#temporal coverage
print("Earliest year:", irena["Year"].min())
print("Latest year:", irena["Year"].max())

#technology and data types
print(irena["Technology"].dropna().unique())
print(irena["Data Type"].dropna().unique())
```
